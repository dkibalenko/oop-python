***class-based***

```python
class Employee:
    bonus_coefficient = 1.1
    
    def __init__(self, base_salary: int):
        self.base_salary = base_salary
    
    def calculate_salary(self):
        return self.base_salary * Employee.bonus_coefficient
```

```python
john.calculate_salary()
```

having overrided parent `__init__` in `User` class, we added `__init__` in the `Admin` class with only `surname` stored in. in this case when instantiate `Admin` class we have only `surname` attribute added. we can either draw `__init__` from parent `User` class using `super().__init__()`, which has `name` attribute and use it in `Admin` or write it directly in the `Admin`

```python
class User:
    role = "user"
    
    def __init__(self, name: str):
        self.name = name
        
    def print_info(self):
        print(f"Name: {self.name}")
        
        
class Admin(User):
    role = "admin"
    
    def __init__(self, name: str, surname: str):
        super().__init__(name)  # we use 'name' inherited from the parent '__init__'
        self.surname = surname
    
    def auth(self):
        print(f"Admin {self.name} authorized")
```

```python
u = User("John")
a = Admin("Eric", "Idle")
```

```python
vars(u)
```

```python
vars(a)
```

```python
dir(a)
```

```python
class User:
    role = "user"
    
    def __init__(self, name: str):
        self.name = name
        
    def print_info(self):
        print(f"Name: {self.name}")
        
        
class Admin(User):
    role = "admin"
    
    def __init__(self, name: str, surname: str):
        super().__init__(name)
        self.surname = surname
    
    def auth(self):
        print(f"Admin {self.name} authorized")
        
    def print_info(self):
        # call parent class `print_info` method
        super(Admin, self).print_info()  # 'super().print_info' is also possible
        print("Hello from Admin!")
```

```python
u = User("John")
a = Admin("Eric", "Idle")
```

### `@classmethod` and `@staticmethod` with inheritance

```python
class Weapon:
    def __init__(self, name: str, power: int):
        self.name = name
        self.power = power
        
    @classmethod
    def from_string_classmethod(cls, weapon_string: str):
        weapon_name, weapon_power = weapon_string.split()
        return cls(
            name=weapon_name,
            power=int(weapon_power)
        )
    
    @staticmethod
    def from_string_staticmethod(weapon_string: str):
        weapon_name, weapon_power = weapon_string.split()
        return Weapon(
            name=weapon_name,
            power=int(weapon_power)
        )
```

```python
class Sword(Weapon):
    weapon = "sword"
```

```python
class Gun(Weapon):
    weapon = "gun"
```

```python
class MachineGun(Weapon):
    weapon = "machine gun"
```

```python
weapon = Weapon.from_string_classmethod("weapon 100")
print(weapon.__class__)
print(weapon.__dict__)
```

```python
weapon = Weapon.from_string_staticmethod("weapon 100")
print(weapon.__class__)
print(weapon.__dict__)
```

```python
sword = Sword.from_string_classmethod("sword 500")
print(sword.__class__)
print(sword.__dict__)
sword.weapon
```

```python
sword = Sword.from_string_staticmethod("sword 500")
print(sword.__class__)
print(sword.__dict__)
sword.weapon
```
