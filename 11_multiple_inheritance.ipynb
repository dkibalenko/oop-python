### Multi-level inheritance

One level inheritance:

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def print_info(self):
        print(f"{self.name} (age:{self.age})")
        
class Staff(User):
    def __init__(self, name, age):
        super().__init__(name, age)
        self.is_staff = True
        
    def login_to_admin_panel(self):
        print(f"{self.name} is entering Admin panel!")
```

```python
john = Staff("John", 25)
```

```python
john.print_info()
```

```python
john.login_to_admin_panel()
```

```python
class Moderator(Staff):
    def __init__(self, name, age):
        super().__init__(name, age)
        self.is_admin = False
        
    def change_article(self):
        print(f"{self.name} changed Article")
```

```python
class Admin(Staff):
    def __init__(self, name, age):
        super().__init__(name, age)
        self.is_admin = True
        
    def create_new_staff(self):
        print(f"{self.name} created new Staff")
```

```python
john = Moderator("John", 30)
```

```python
john.print_info()
```

```python
john.login_to_admin_panel()
```

```python
john.change_article()
```

```python
try:
    john.create_new_staff()
except AttributeError as ex:
    print(ex)
```

```python
eric = Admin('Eric', 27)
```

```python
eric.print_info()
```

```python
eric.login_to_admin_panel()
```

```python
eric.create_new_staff()
```

```python
try:
    eric.chage_article()
except AttributeError as ex:
    print(ex)
```

### Multiple inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name
        
    def eat(self):
        print(f"{self.name} eats")
        
class Cat(Animal):
    def sleep(self):
        print(f"{self.name} sleeps")
        
class Dog(Animal):
    def bark(self):
        print(f"{self.name} barks")
```

```python
class CatDog(Cat, Dog):
    def laugh(self):
        print(f"{self.name} laughs")
```

```python
cat_dog = CatDog("CatDog")
```

```python
cat_dog.eat()
cat_dog.sleep()
cat_dog.bark()
cat_dog.laugh()
```

```python
dir(cat_dog)
```

### MRO

```python
class Animal:
    def __init__(self, name):
        self.name = name
        
    def eat(self):
        print(f"{self.name} eats")
        
class Cat(Animal):
    def sleep(self):
        print(f"{self.name} sleeps")
        
    def eat(self):
        print(f"{self.name} drinks milk")
        
class Dog(Animal):
    def bark(self):
        print(f"{self.name} barks")
        
    def eat(self):
        print(f"{self.name} eats steak")
        
class CatDog(Cat, Dog):
    def laugh(self):
        print(f"{self.name} laughs")
```

```python
cat_dog = CatDog("CatDog")
```

```python
cat_dog.eat()
```

```python
cat_dog.sleep()
```

```python
CatDog.mro()
```

The ***method resolution order*** (or MRO) tells Python how to search for inherited methods. This comes in handy when you’re using `super()` because the MRO tells you exactly where Python will look for a method you’re calling with `super()` and in what order.

Every class has an `.__mro__` attribute that allows us to inspect the order.

```python
class Animal:
    def __init__(self, name):
        self.name = name
        
    def eat(self):
        print(f"{self.name} eats")
        
class Cat(Animal):
    def sleep(self):
        print(f"{self.name} sleeps")
        
#     def eat(self):
#         print(f"{self.name} drinks milk")
        
class Dog(Animal):
    def bark(self):
        print(f"{self.name} barks")
        
    def eat(self):
        print(f"{self.name} eats steak")
        
class CatDog(Cat, Dog):
    def laugh(self):
        print(f"{self.name} laughs")
```

let's now change `Cat` and see which class `eat()` is now going to be accessed:

```python
cat_dog = CatDog("CatDog")
```

```python
cat_dog.eat()
```

now it's clear, `eat()` is taken from `Dog`

```python
CatDog.mro()
```

it went to `Cat` first, didn't find it, then next by MRO is `Dog` which has this method

if none of the classes in the inheritance chain has `eat()`, an exception would be going to be raised:

```python
class Animal:
    def __init__(self, name):
        self.name = name
        
#     def eat(self):
#         print(f"{self.name} eats")
        
class Cat(Animal):
    def sleep(self):
        print(f"{self.name} sleeps")
        
#     def eat(self):
#         print(f"{self.name} drinks milk")
        
class Dog(Animal):
    def bark(self):
        print(f"{self.name} barks")
        
#     def eat(self):
#         print(f"{self.name} eats steak")
        
class CatDog(Cat, Dog):
    def laugh(self):
        print(f"{self.name} laughs")
```

### Multiple Inheritance Example

```python
class Transport:
    def __init__(self, position):
        self.position = position
        
    def travel(self, destination):
        print(f"Travel from {self.position} to {destination}")
        self.position = destination
        
class Car(Transport):
    pass

class Boat(Transport):
    pass

class Plane(Transport):
    pass
```

```python
car = Car("Kyiv")
```

```python
car.travel("Lviv")
```

```python
boat = Boat("Left Bank")
boat.travel("Right Bank")
```

we decided to add method to `Car`:

```python
class Transport:
    def __init__(self, position):
        self.position = position
        
    def travel(self, destination):
        print(f"Travel from {self.position} to {destination}")
        self.position = destination
        
class Car(Transport):
    @staticmethod
    def play_song_on_station(station):
        print(f"play song on station {station}")

class Boat(Transport):
    pass

class Plane(Transport):
    pass
```

```python
car = Car("Kyiv")
```

```python
car.play_song_on_station("FM")
```

```python
class Phone:
    @staticmethod
    def play_song_on_station(station):
        print(f"play song on station {station}")
```

```python
phone = Phone()
```

```python
phone.play_song_on_station("Radio Forever")
```

### Operator Overloading

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
```

```python
dog1 = Dog("Name1", "Breed1")
dog2 = Dog("Name2", "Breed2")
```

```python
try:
    dog1 < dog2
except TypeError as ex:
    print(ex)
```

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
        
    def __lt__(self, other):
        return self.name < other.name
```

```python
dog1 = Dog("Name1", "Breed1")
dog2 = Dog("Name2", "Breed2")
```

```python
dog1 < dog2
```
