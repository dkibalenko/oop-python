### Encapsulation

The concept of wrapping data and methods that work with data in one unit. It prevents data modification accidentally by limiting access to variables and methods

You can use three forms of access modifiers to implement attributes and methods:
* public access modifier;
* protected access modifier;
* private access modifier.

***Public Access Modifier***

The class members declared ***public*** are easily accessible from any part of the program. All attributes and methods of a class are public by default.

```python
class Base:
    def __init__(self):
        self.public = "public"

    def print_info(self):
        print(f"{self.public} attribute")


class Derived(Base):
    def __init__(self):
        super().__init__()
        print(self.public)


base = Base()

print(base.public)        # public
base.print_info()         # public attribute
```

***Protected Access Modifier***

The class members declared ***protected*** are only accessible to a class inherited from it and, of course, in a class itself. Use a single underscore `_` symbol before the data to declare it protected. You can still access ***protected*** attributes, and methods outside the class and values will be accessed, but it is a bad practice.

```python
class Base:
    def __init__(self):
        self._protected = "protected"

    def print_info(self):
        print(f"{self._protected} attribute")


class Derived(Base):
    def __init__(self):
        super().__init__()
        print(self._protected)


base = Base()

print(base._protected)        # protected
base.print_info()             # protected attribute
```

```python
derived = Derived()
```

```python
derived._protected  # bad practice
```

***Private Access Modifier***

The class members declared ***private*** are accessible within the class only. Use a double underscore `__` symbol before the data to declare it private.

```python
class Base:
    def __init__(self):
        self.__private = "private"

    def __print_info(self):
        print(f"{self.__private} attribute")


class Derived(Base):
    def __init__(self):
        super().__init__()
        print(self.__private)


base = Base()

try:
    base.__private  # Error
except AttributeError as ex:
    print(ex)
    
try:
    base.__print_info()  # Error
except AttributeError as ex:
    print(ex)
    
try:
    Derived()  # Error
except AttributeError as ex:
    print(ex)
```

In the example above, `__private` and `__print_info()` are ***private*** members, so they can be accessed only inside the `Base` class.

Although it is a bad practice, but you can access private attributes outside of a class using this syntax:

```python
class Base:
    def __init__(self):
        self.__private = "private"


base = Base()

print(base._Base__private) # private
```

#### Public vs Protected vs Privat

```python
class Base:
    def __init__(self):
        self.zero = "public"
        
        self._first = "protected"
        self.__second = "private"
        
    def public_method(self):
        # ok practice to use protected & private inside class
        return self.zero + self._first + self.__second
    
    def _protected(self):
        print("call _protected")
        
    def __private(self):
        print("call __private")
```

```python
base = Base()
```

by convention, we should use attributes as follows:
* 1. Inside class
* 2. Inside Child class
* 3. Outside class

we can use attributes:
* ***public*** `->` ***everywhere***
* ***protected*** `->` inside ***class*** or ***child class***
* ***private*** `->` inside ***class***

good practice:

```python
base.public_method()
```

bad practice:

```python
base._first = "new protected"
base._protected()
```

get an ***Error***:

```python
base.__second
```

we still can access ***private*** attibute:

```python
base._Base__second = "new private"
base._Base__private()
```

but this is not a good practice

let see at a ***child class***:

```python
class Derived(Base):
    def __init__(self):
        super().__init__()  # a good practice to access 'protected' attibutes of the 'parent' class
        
        # ok practice:
        print("Protected member of Base class:", self._first)
        self._protected()

        # bad practice:
        print("Private member of Base class:", self.__second)
        self.__private()
```

```python
base.public_method()
```

### Abstraction

The OOP concept that shows only essential attributes and hides unnecessary information. The main purpose of abstraction is to hide unnecessary details from the users. Abstraction selects data from a larger pool to show only relevant details of the object to the user. It helps in reducing programming complexity and efforts.

```python
class Cafe:
    def __init__(self, name):
        self.name = name
        self._workers = ["John", "Kate", "Bob"]
        
        
    def open(self):
        print("Cafe is working!")
        
        
    def order_coffee(self):
        if len(self._workers) > 0:
            print("Making coffee...")
            print("Please, here you are!")
        else:
            print("Error!")
            
    def _sick_leave(self, worker):
        self._workers.remove(worker)
        
    def _skip_job(self, worker):
        self._workers.remove(worker)
        
    def _ask_overtime(self, different_worker):
        self._workers.append(different_worker)
```

```python
best_cafe = Cafe("The best ever Coffee")
```

```python
best_cafe.open()
```

```python
best_cafe.order_coffee()
```

### ABC, `@abstractmethod`

Abstract classes are classes that contain one or more abstract methods. The concrete class provides an implementation of abstract methods, the abstract base class can also provide an implementation by invoking the methods via super(). Abstract classes cannot be instantiated and require subclasses to provide implementations for abstract methods.

```python
class Animal:
    def eat(self):
        pass
    
    
class Tiger(Animal):
    def eat(self):
        print("Tiger eats steak")
        
        

class Zebra(Animal):
    def eat(self):
        print("Zebra eats grass")
        
        
        
class Elephant(Animal):
    def elephant_eat(self):
        print("Elephant eats peanuts")
        
        
        
class Penguin(Animal):
    def eating(self):
        print("Penguin is eating fish")
```

```python
animals = [Tiger(), Zebra(), Elephant(), Penguin()]
```

```python
for animal in animals:
    animal.eat()
```

by defining `@abstractmethod` for ***parent*** method, we say that the method had better off be redefined in all child classes:

```python
from abc import abstractmethod
```

```python
class Animal:
    @abstractmethod
    def eat(self):
        pass
    
    
class Tiger(Animal):
    def eat(self):
        print("Tiger eats steak")
        
        

class Zebra(Animal):
    def eat(self):
        print("Zebra eats grass")
        
        
        
class Elephant(Animal):
    def eat(self):
        print("Elephant eats peanuts")
        
        
        
class Penguin(Animal):
    def eating(self):
        print("Penguin is eating fish")
```

```python
animals = [Tiger(), Zebra(), Elephant(), Penguin()]

for animal in animals:
    animal.eat()
```

we still missed to redefine method for `Penguin`, but we would have liked to know it before for all child classes, then we can use `ABC` - ***abstract base class***:

```python
from abc import ABC
```

```python
class Animal(ABC):
    @abstractmethod
    def eat(self):
        pass
    
    
class Tiger(Animal):
    def eat(self):
        print("Tiger eats steak")
        
        

class Zebra(Animal):
    def eat(self):
        print("Zebra eats grass")
        
        
        
class Elephant(Animal):
    def eat(self):
        print("Elephant eats peanuts")
        
        
        
class Penguin(Animal):
    def eating(self):
        print("Penguin is eating fish")
```

```python
animals = [Tiger(), Zebra(), Elephant(), Penguin()]

for animal in animals:
    animal.eat()
```

now, we cannot create an instance of the class `Penguin` unless we define `eat()` method in there

this exception is raised when we define a class with no such parent method implemented

acctually we now ***must*** to implement `eat()` method for all child classes

```python
class Animal(ABC):
    @abstractmethod
    def eat(self):
        pass
    
    
class Tiger(Animal):
    def eat(self):
        print("Tiger eats steak")
        
        

class Zebra(Animal):
    def eat(self):
        print("Zebra eats grass")
        
        
        
class Elephant(Animal):
    def eat(self):
        print("Elephant eats peanuts")
        
        
        
class Penguin(Animal):
    def eat(self):
        print("Penguin is eating fish")
```

```python
animals = [Tiger(), Zebra(), Elephant(), Penguin()]

for animal in animals:
    animal.eat()
```
